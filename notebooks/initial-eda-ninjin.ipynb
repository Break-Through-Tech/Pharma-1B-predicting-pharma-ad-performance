### Which columns exist at prediction time?
For forecasting a campaign that hasn't run yet, we only know the **pre-campaign**
columns. Everything measured after the campaign ran is an *outcome*, not a feature:

- **Known upfront (inputs):** brand, sub_product_name, country, pillar, media_type,
  profession, npi_specialty, region, start date, cpe, pre_nrx, pre_avg_nrx
- **Measured afterward (outcomes):** reach, engagement columns,
  total_budget_spent, total_spend, post_nrx, nrx_lift, nrx_lift_pct
- **Never use:** actual_spend_per_engagement, spend_per_nrx_lift (arithmetic
  restatements of the target)

```python
# Decode profession codes
PRACTITIONER_CATEGORY = {
    "1": "Clinical Psychologist", "2": "General Psychologist", "3": "Social Worker",
    "4": "Occupational Therapist", "5": "Mental Health Nurse",
    "6": "Aboriginal and Torres Strait Islander Health/Mental Health Worker",
    "7": "Low Intensity Mental Health Worker", "8": "General Practitioner",
    "9": "Psychiatrist", "10": "Other Medical", "11": "Other",
    "12": "Psychosocial Support Worker", "13": "Peer Support Worker", "99": "Not stated",
}

prof = df["profession"].astype("string").str.strip()
df["profession_label"] = prof.map(PRACTITIONER_CATEGORY).fillna(prof)

# numeric codes that aren't in the mapping
is_code = prof.str.fullmatch(r"\d+").fillna(False)
unmapped = is_code & ~prof.isin(PRACTITIONER_CATEGORY.keys())
df.loc[unmapped, "profession_label"] = "Unmapped code"

# the 56 structural nulls (push tactics with no on-site targeting)
df["profession_label"] = df["profession_label"].fillna("not_applicable")

print(df["profession_label"].value_counts())
print("\nunmapped codes:", prof[unmapped].value_counts().to_dict())
print("plain text entries:", "General Practitioner,", "psychologist,", "physician assistant,", "physician,", "Psychological Support Worker,", "Peer Support Worker")
```

```python
# Clarify budget columns
# total_spend = total budget available for the whole campaign run
# total_budget_spent = amount spent so far on launching tactics at a certain time

df["budget_utilization"] = df["total_budget_spent"] / df["total_spend"].replace(0, np.nan)

print("rows where spent > budget:", (df.total_budget_spent > df.total_spend).sum())
print(df["budget_utilization"].describe(percentiles=[.1, .25, .5, .75, .9]).round(5))
```
